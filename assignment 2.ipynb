```python
def rotate_right_without_carry(binary_str, positions):
    """Performs Rotate Right Without Carry (ROR) by n positions."""
    print(f"\n--- ROR on {binary_str} by {positions} positions ---")
    print(f"Original 8-bit number : {binary_str}")
    
    n = len(binary_str)
    positions = positions % n
    
    # Step-by-step demonstration
    current = binary_str
    for i in range(1, positions + 1):
        # Move the last bit to the front
        current = current[-1] + current[:-1]
        print(f"Step {i} (Shift right): {current}")
        
    print(f"Final 8-bit result    : {current}")
    return current

def rotate_right_with_carry(binary_str, positions, initial_carry=0):
    """Performs Rotate Right With Carry (RCR) using 9 bits (8-bit value + Carry flag)."""
    print(f"\n--- RCR on {binary_str} by {positions} positions (Initial Carry = {initial_carry}) ---")
    print(f"Original 8-bit number : {binary_str}")
    print(f"Initial Carry Flag    : {initial_carry}")
    
    carry = initial_carry
    current = list(binary_str)
    
    for i in range(1, positions + 1):
        # In RCR, the rightmost bit goes to Carry, 
        # and the old Carry goes to the leftmost position of the 8-bit number.
        old_rightmost = current[-1]
        current = [str(carry)] + current[:-1]
        carry = int(old_rightmost)
        
        print(f"Step {i} -> Number: {''.join(current)}, New Carry: {carry}")
        
    final_result = ''.join(current)
    print(f"Final 8-bit result    : {final_result}")
    print(f"Final Carry Flag      : {carry}")
    return final_result, carry

# a) Rotate right without carry (ROR) on 11011101 by 3 positions
rotate_right_without_carry("11011101", 3)

# b) Rotate right with carry (RCR) on 11010001 by 3 positions 
# (Assuming initial Carry = 0 unless specified otherwise)
rotate_right_with_carry("11010001", 3, initial_carry=0)
```
