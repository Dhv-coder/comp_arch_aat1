```python
def to_twos_complement_8bit(decimal_val):
    """Converts a decimal integer to its 8-bit 2's complement binary string."""
    if decimal_val < -128 or decimal_val > 127:
        raise ValueError(f"Value {decimal_val} is out of 8-bit signed range (-128 to 127).")
    
    if decimal_val < 0:
        binary_val = bin((1 << 8) + decimal_val)[2:]
    else:
        binary_val = bin(decimal_val)[2:].zfill(8)
    return binary_val

def from_twos_complement(binary_str):
    """Converts an 8-bit 2's complement binary string back to a decimal integer."""
    if binary_str[0] == '1':
        # Negative number: invert bits and add 1, then apply negative sign
        inverted = ''.join('1' if b == '0' else '0' for b in binary_str)
        decimal_val = -(int(inverted, 2) + 1)
    else:
        decimal_val = int(binary_str, 2)
    return decimal_val

def perform_signed_arithmetic(num1, num2, operation):
    """
    Performs signed arithmetic using 8-bit 2's complement representation,
    checks for overflow, and returns the result in binary and decimal.
    """
    print(f"\n--- Expression: {num1} {operation} ({num2} if num2 < 0 else num2) ---")
    
    # Adjust second number based on operation (+ or -)
    actual_num2 = -num2 if operation == '-' else num2
    
    print(f"Operand 1 ({num1}): {to_twos_complement_8bit(num1)}")
    print(f"Operand 2 ({num2 if operation == '+' else -num2}): {to_twos_complement_8bit(actual_num2)} (after applying operator '{operation}')")
    
    # Perform addition/subtraction in decimal to check true value and overflow
    if operation == '+':
        result_decimal = num1 + num2
    elif operation == '-':
        result_decimal = num1 - num2
    else:
        raise ValueError("Unsupported operation")
        
    # Check for 8-bit signed overflow (-128 to 127)
    overflow = False
    if result_decimal < -128 or result_decimal > 127:
        overflow = True
        
    print(f"Decimal Result: {result_decimal}")
    if overflow:
        print("⚠️ Arithmetic Overflow detected! The result exceeds 8-bit signed limits.")
    else:
        binary_result = to_twos_complement_8bit(result_decimal)
        print(f"Binary Result (8-bit 2's complement): {binary_result}")

# Executing the expressions given in the question:
# a) +56 + (+29)
perform_signed_arithmetic(56, 29, '+')

# b) (-38) - (-23)
perform_signed_arithmetic(-38, -23, '-')
```
